```python
#Simple Calculator Using Python Functions


#Functionn for addition

def add(a,b):
  return a + b

#Function for subtraction

def subtract(a,b):
  return a - b

#Fucntion for multliplication

def multiply(a,b):
  return a * b

#Function for division

def divide(a,b):

  #check for division by zero
  if b == 0:

    return "Error: Cannot divide by zero."
    return a/b
#Function for modulus

def mod(a,b):
  return a % b

#Function for power

def pow(a,b):
  return a ** b


#Display calculator options

print("Simple Calculator")
print("------------")
print("1. Addition")
print("2. Subtraction")
print("3. Mulitplicaiton")
print("4. Division")
print("5. Modulus")
print("6. Power")

#Get input from the user

choice = input("Choose and operation (1 - 6): ")

num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

#Perform the selected operation

if choice == "1":
  result = add(num1, num2)
  print("Result:", result)

elif choice == "2":
  result = subtract(num1, num2)
  print("Result:", result)

elif choice == "3":
  result = multiply(num1, num2)
  print("Result:", result)

elif choice == "4":
  result = divide(num1, num2)
  print("Result:", result)

elif choice == "5":
  result = mod(num1, num2)
  print("Result:", result)

elif choice == "6":
  result = pow(num1, num2)
  print("Result:", result)

else:
  print("Invalid choice.")






```
